```python
import numpy as np
```

```python
arr = [47,23,7,17,63,34]
A = np.array(arr)
print("Numpy Array: ",A)
```

```python
print("Array: ",A)
print("A[0:3]: ",A[0:3])
print("A[1:]: ",A[1:])
print("A[-5:-3]: ",A[-5:-3])
print("A[-2:]: ",A[-2:])
```

```python
print("Array: ",A)
print("Addition: ",A+5)
print("Subtraction: ",A-5)
print("Multiplication: ",A*5)
print("Division: ",A/5)
```

```python
print("Array: ",A)
print("Sum: ",np.sum(A))
print("Mean: ",np.mean(A))
print("Max: ",np.max(A))
print("Min: ",np.min(A))
```

```python
print("Array: ",A)
print("A(2X3): \n ",A.reshape(2,3))
print("A(3X2): \n",A.reshape(3,2))
```
