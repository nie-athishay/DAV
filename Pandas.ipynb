```python
import pandas as pd
```

```python
data = {
    "Name" : ["Akash","Manu","Santhosh"],
    "Marks" : [7,17,23],
    "Branch" : ["CSE","ECE","EEE"]
}
print(data)
```

```python
df = pd.DataFrame(data)
print(df)
```

```python
#Column Indexing
for x in data:
    print(x)
    print(df[x])
    
```

```python
#Row Indexing using iloc
n = len(data)
for i in range(n):
    print(df.iloc[i])
```
